## 4. Modelos Baseline (Naive)

Los modelos naive sirven como referencia mínima — cualquier modelo
más complejo debe superarlos para justificar su uso.

**Naive_last:** predice el último valor observado para todo el horizonte.

**Naive_seasonal:** retrocede exactamente H pasos atrás y copia ese
segmento como predicción:
- Para 48h -> copia los 2 días anteriores
- Para 72h -> copia los 3 días anteriores

```python
def forecast_naive_last(train, h):
    return np.repeat(train.values[-1], h)

def forecast_naive_seasonal(train, h, season=96):
    """
    Predice replicando el patrón del día anterior.
    Para cada paso h, copia el valor de hace exactamente 'season' pasos.
    """
    if len(train) < season:
        return forecast_naive_last(train, h)
    # Para cada paso futuro i, copiar train[-(season - i % season)]
    result = np.empty(h)
    for i in range(h):
        result[i] = train.values[-(season - (i % season))]
    return result

print("Naive_last: predice el último valor observado")
print("Naive_seasonal: copia el patrón de hace H pasos")
```

## 7. Selección del orden ARIMA — Metodología Box-Jenkins

La metodología Box-Jenkins define un proceso sistemático para
identificar, estimar y validar modelos ARIMA:

1. **Identificación**: análisis de ACF y PACF para determinar p, d, q
   - PACF corta en lag p -> componente AR(p)
   - ACF corta en lag q -> componente MA(q)
   - Número de diferenciaciones necesarias para estacionariedad -> d

2. **Estimación**: ajuste del modelo por máxima verosimilitud

3. **Validación**: análisis de residuos — deben comportarse como ruido blanco

En este notebook se automatiza la fase de identificación con
auto_arima, que implementa la búsqueda stepwise minimizando el AIC.
Esto es equivalente al proceso manual Box-Jenkins pero de forma
sistemática y reproducible.

```python
from pmdarima import auto_arima

# Usar últimos 90 días de train para seleccionar orden
sample = y_train.iloc[-90 * SEASONAL_PERIOD:]

print(f"Muestra para auto_arima: {len(sample)} observaciones ({len(sample)//96} días)")
print("Buscando orden óptimo con auto_arima...")
auto_model = auto_arima(
    sample.values,
    start_p=0, max_p=8,
    start_q=0, max_q=4,
    d=None,              # auto-detectar diferenciación
    seasonal=False,      # sin componente estacional 
    information_criterion="aic",
    stepwise=True,
    suppress_warnings=True,
    error_action="ignore",
    verbose=False
)

arima_order = auto_model.order
print("Orden óptimo encontrado:", arima_order)
print("AIC:", round(auto_model.aic(), 2))
print(auto_model.summary())
```

## 8. Interpretación del orden seleccionado

auto_arima selecciona **ARIMA(6,1,1)** como orden óptimo sobre los últimos 90 días de entrenamiento:

- **d=1**: la serie requiere una diferenciación para estabilizar la media local
- **p=6**: el modelo utiliza las 6 horas anteriores como predictores autorregresivos
- **q=1**: se incluye un término de media móvil

Los coeficientes AR muestran una dependencia positiva fuerte con el lag 1 (1.28), un efecto corrector negativo en el lag 2 (-0.77) y una señal significativa en el lag 6 (-0.16), que corresponde a un cuarto del ciclo diario. El coeficiente ar.L5 no es estadísticamente significativo (p=0.156), pero se mantiene porque ar.L6 sí lo es.

El test de Ljung-Box (p=0.68) confirma que los residuos no presentan autocorrelación significativa. El test de Jarque-Bera rechaza la normalidad de los residuos (curtosis=7.89), lo que indica la presencia de valores extremos, esperable en datos de huella de carbono con eventos atípicos del sistema eléctrico.

A pesar de este ajuste razonable sobre datos recientes, el ARIMA carece de componente estacional explícita, por lo que se espera un rendimiento limitado en horizontes largos (48–72h) donde el patrón diario es determinante.

## 9. Evaluación ARIMA con walk-forward

Se evalúa ARIMA con ~50 fits distribuidos semanalmente por 2024.
En cada fit el modelo se reentrena con todo el histórico disponible
hasta ese momento — ventana expansiva.

```python
from statsmodels.tsa.arima.model import ARIMA

def forecast_arima(train, h, order):
    
    model = ARIMA(
        train.values,
        order=order,
        trend="n",                   # sin intercepto 
        enforce_stationarity=False,
        enforce_invertibility=False
    )
    
    res = model.fit(method_kwargs={"warn_convergence": False})
    
    return res.forecast(steps=h)
```

```python
print(f"Evaluando ARIMA{arima_order} con ~50 evaluaciones semanales...")
arima_val, preds_arima = evaluate_model(
    y_train, y_val, HORIZONS, cfg_arima,
    lambda tr, h: forecast_arima(tr, h, arima_order),
    verbose=True
)

arima_summary = summarize(arima_val, f"ARIMA{arima_order}")

# Con intervalos de confianza
arima_ci = summarize_with_ci(arima_val, f"ARIMA{arima_order}")

print("\n=== Resultados con IC 95% ===")
for _, r in arima_ci.iterrows():
    print(f"{r['model']:20s} {r['horizon']:4s}  "
          f"MAE = {r['MAE_mean']:.2f} ± {r['MAE_std']:.2f}  "
          f"[{r['MAE_ci_lo']:.2f}, {r['MAE_ci_hi']:.2f}]  "
          f"(n={int(r['n_eval'])})")
```

```python
def build_results_row(summary_df):
    """Convierte un summarize() en una fila con MAE y RMSE por horizonte."""
    row = {"model": summary_df["model"].iloc[0]}
    for _, r in summary_df.iterrows():
        row[f"MAE_{r['horizon']}"]  = round(r["MAE"], 3)
        row[f"RMSE_{r['horizon']}"] = round(r["RMSE"], 3)
    return row

resultados = pd.DataFrame([
    build_results_row(summarize(naive_last_val, "Naive_last")),
    build_results_row(summarize(naive_seas_val, "Naive_seasonal")),
    build_results_row(arima_summary),
]).set_index("model")

print(resultados)
resultados.to_csv(RESULTS_DIR / "comparacion_naive_arima.csv")
```

## Interpretación de resultados — Naive y ARIMA

### Naive_last
El modelo de persistencia simple obtiene un MAE de 28.72 gCO₂eq/kWh a 48h y 30.66 a 72h. Es el peor modelo de los evaluados, lo que confirma que la serie tiene estructura temporal explotable — el último valor observado no es una buena estimación del futuro.

### Naive_seasonal
El modelo estacional obtiene un MAE de 18.24 gCO₂eq/kWh a 48h y 19.61 a 72h, mejorando un 36% sobre Naive_last. Este resultado confirma que el patrón diario es la característica dominante de la serie — copiar las últimas 48 o 72 horas es una estrategia competitiva que cualquier modelo más complejo debe superar.

Este modelo se establece como **baseline principal** del TFM.

### ARIMA(6,1,1)
ARIMA obtiene un MAE de 25.61 a 48h y 27.88 a 72h. Mejora ligeramente respecto a Naive_last (~11%) pero queda muy por encima del Naive_seasonal.

Las gráficas de predicción ilustran la causa: la predicción ARIMA converge en pocas horas a un valor plano cercano a la media local, perdiendo completamente las oscilaciones diarias. Con solo 6 lags horarios de memoria y sin componente estacional, el modelo es incapaz de proyectar el ciclo de 24 horas que domina la serie.

| Modelo | MAE 48h | MAE 72h | RMSE 48h | RMSE 72h |
|--------|---------|---------|----------|----------|
| Naive_last | 28.72 | 30.66 | 34.94 | 36.98 |
| **Naive_seasonal** | **18.24** | **19.61** | **22.60** | **24.49** |
| ARIMA(6,1,1) | 25.61 | 27.88 | 31.72 | 33.89 |

Este resultado motiva el uso de modelos con mayor capacidad de memoria
y captura de dependencias temporales de largo alcance. En particular,
el modelo AutoReg con un elevado número de rezagos permite capturar
el patrón diario completo sin el elevado coste computacional de los
modelos SARIMA con periodo estacional s=96, que fueron descartados
por inviabilidad computacional en el contexto de evaluación walk-forward.
Posteriormente se implementan modelos de aprendizaje profundo (LSTM)
para explorar la captura de dependencias no lineales.
